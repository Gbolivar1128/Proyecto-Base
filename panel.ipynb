```python
import json
```

```python
import pandas as pd
```

```python
with open("data.json", "r", encoding="utf-8") as f:
    datos = json.load(f)

df = pd.DataFrame(datos["ventas"])
df["total"] = df["cantidad"] * df["precio_unitario"]
```

```python
ingresos_por_producto = df.groupby("producto")["total"].sum()
max_ingresos = ingresos_por_producto.max()
productos_top = ingresos_por_producto[ingresos_por_producto == max_ingresos].index.tolist()
print(f"Producto(s) con más ingresos ({max_ingresos:.2f} €): {', '.join(productos_top)}")
```

```python
vendedor_concreto = "Marta"
ventas_vendedor = df[df["vendedor"] == vendedor_concreto]
display(ventas_vendedor)
```

```python
ticket_medio = df["total"].mean()
print(f"Ticket medio: {ticket_medio:.2f} €")
```
